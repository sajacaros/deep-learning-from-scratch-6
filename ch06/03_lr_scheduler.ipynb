# 03_lr_scheduler

『밑바닥부터 시작하는 딥러닝 ❻』 실습 코드 — 원본: `ch06/03_lr_scheduler.py`

셀을 위에서부터 차례대로 실행하세요.

```python
def get_lr(it, max_lr, warmup_iters, max_iters):
    # 워밍업： 0 -> max_lr
    if it < warmup_iters:
        return max_lr * (it / warmup_iters)

    # 어닐링： max_lr -> 0
    if it < max_iters:
        progress = (it - warmup_iters) / (max_iters - warmup_iters)
        return max_lr * (1.0 - progress)

    return 0.0
```
